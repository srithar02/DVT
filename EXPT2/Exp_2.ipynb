```python
import pandas as pd
import os

# ---------- Import the complete dataset from CSV ----------
df = pd.read_csv("popular_anime.csv")

print("Shape   :", df.shape)
print("Columns :", list(df.columns))
```

```python
print(df.head(5)[["id", "name", "type", "episodes", "score"]])
print()
print("Memory used :", round(df.memory_usage(deep=True).sum() / 1024**2, 2), "MB")
```

```python
# ---------- Import only the required columns ----------
cols = pd.read_csv("popular_anime.csv", usecols=["name", "type", "score", "scored_by"])

print("Shape with selected columns :", cols.shape)
print(cols.head())
```

```python
# ---------- Import a limited number of rows ----------
top = pd.read_csv("popular_anime.csv", nrows=8, usecols=["name", "score", "rank"])

print("Shape with nrows = 8 :", top.shape)
print(top)
```

```python
# ---------- Import the file in chunks ----------
total = 0
for i, chunk in enumerate(pd.read_csv("popular_anime.csv", chunksize=10000), start=1):
    total += len(chunk)
    print("Chunk", i, "-> rows :", len(chunk), " running total :", total)

print("\nAll chunks processed. Total rows =", total)
```

```python
# ---------- Create a working subset for export ----------
subset = (df.dropna(subset=["score"])
            .sort_values("score", ascending=False)
            .head(50)[["id", "name", "type", "episodes", "score", "scored_by"]])

print("Subset shape :", subset.shape)
print(subset.head(3))
```

```python
# ---------- Export to CSV ----------
subset.to_csv("anime_top50.csv", index=False)
print("CSV file written.")

# ---------- Export to Excel ----------
subset.to_excel("anime_top50.xlsx", index=False, sheet_name="Top50")
print("Excel file written.")

# ---------- Export to JSON ----------
subset.to_json("anime_top50.json", orient="records", indent=2)
print("JSON file written.")

# ---------- Export to pickle ----------
subset.to_pickle("anime_top50.pkl")
print("Pickle file written.")
```

```python
# ---------- Read the exported files back to verify ----------
from_csv   = pd.read_csv("anime_top50.csv")
from_excel = pd.read_excel("anime_top50.xlsx")
from_json  = pd.read_json("anime_top50.json")
from_pkl   = pd.read_pickle("anime_top50.pkl")

print("CSV   :", from_csv.shape)
print("Excel :", from_excel.shape)
print("JSON  :", from_json.shape)
print("Pickle:", from_pkl.shape)
print("\nAll four exports match the original subset :",
      all(len(x) == len(subset) for x in [from_csv, from_excel, from_json, from_pkl]))
```

```python
# ---------- List the exported files with their sizes ----------
for f in ["anime_top50.csv", "anime_top50.xlsx", "anime_top50.json", "anime_top50.pkl"]:
    print(f.ljust(20), "->", round(os.path.getsize(f) / 1024, 2), "KB")

print()
print(from_csv.head(3).to_string(index=False))
```
