```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv("popular_anime.csv")
df = df[["name", "type", "episodes", "status", "score",
         "scored_by", "rank", "rating", "studios"]]
print("Shape :", df.shape)
```

```python
# ---------- Inspection : first and last records ----------
print("First 5 records :")
print(df.head()[["name", "type", "episodes", "score"]])
print("\nLast 3 records :")
print(df.tail(3)[["name", "type", "episodes", "score"]])
```

```python
# ---------- Inspection : structure of the dataset ----------
df.info()
```

```python
# ---------- Inspection : data types, unique values and missing values ----------
summary = pd.DataFrame({
    "dtype"   : df.dtypes.astype(str),
    "unique"  : df.nunique(),
    "missing" : df.isna().sum()
})
print(summary)
```

```python
# ---------- Inspection : descriptive summary ----------
print(df.describe().round(3))
```

```python
# ---------- Filtering : single condition ----------
high = df[df["score"] > 8.5]
print("Titles scoring above 8.5 :", len(high))
print(high[["name", "type", "score"]].head())
```

```python
# ---------- Filtering : multiple conditions ----------
picks = df[(df["type"] == "TV") & (df["score"] > 8.0) & (df["episodes"] <= 26)]

print("TV series, score above 8.0, at most 26 episodes :", len(picks))
print(picks[["name", "episodes", "score"]].head())
```

```python
# ---------- Filtering : query, isin and between ----------
q = df.query("score >= 8.7 and scored_by > 500000")
print("Using query() :", len(q))
print(q[["name", "score", "scored_by"]].head(3))

print("\nUsing isin() :", len(df[df["type"].isin(["Movie", "OVA"])]))
print("Using between() :", len(df[df["score"].between(7.0, 7.5)]))
```

```python
# ---------- Selection using loc and iloc ----------
print("Using loc :")
print(df.loc[df["score"] > 9.0, ["name", "type", "score"]])

print("\nUsing iloc (rows 0-4, columns 0-3) :")
print(df.iloc[0:5, 0:4])
```

```python
# ---------- Statistics : measures of central tendency ----------
s = df["score"].dropna()

print("Mean   :", round(s.mean(), 4))
print("Median :", round(s.median(), 4))
print("Mode   :", round(s.mode()[0], 4))
```

```python
# ---------- Statistics : measures of dispersion ----------
print("Minimum   :", s.min())
print("Maximum   :", s.max())
print("Range     :", round(s.max() - s.min(), 3))
print("Variance  :", round(s.var(), 4))
print("Std. dev. :", round(s.std(), 4))
print("Q1        :", s.quantile(0.25))
print("Q3        :", s.quantile(0.75))
print("IQR       :", round(s.quantile(0.75) - s.quantile(0.25), 3))
```

```python
# ---------- Statistics : measures of shape ----------
print("Skewness :", round(s.skew(), 4))
print("Kurtosis :", round(s.kurt(), 4))

if s.skew() < 0:
    print("\nThe score distribution is negatively skewed.")
else:
    print("\nThe score distribution is positively skewed.")
```

```python
# ---------- Correlation matrix ----------
num = df[["episodes", "score", "scored_by", "rank"]]
corr = num.corr()

print(corr.round(4))
```

```python
# ---------- Strongest correlations ----------
pairs = corr.where(~np.eye(len(corr), dtype=bool)).stack().sort_values()

print("Strongest negative correlation :")
print(" ", pairs.index[0], "->", round(pairs.iloc[0], 4))
print("Strongest positive correlation :")
print(" ", pairs.index[-1], "->", round(pairs.iloc[-1], 4))
```

```python
# ---------- Correlation heatmap ----------
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(range(len(corr)), corr.columns, rotation=45, ha="right")
ax.set_yticks(range(len(corr)), corr.columns)

for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, round(corr.iloc[i, j], 2), ha="center", va="center", fontsize=10)

ax.set_title("Correlation Matrix of Numeric Columns")
fig.colorbar(im, shrink=0.8)
plt.tight_layout()
plt.show()
```
