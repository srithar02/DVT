```python
import numpy as np
import pandas as pd
import matplotlib
import scipy

print("NumPy      :", np.__version__)
print("Pandas     :", pd.__version__)
print("Matplotlib :", matplotlib.__version__)
print("SciPy      :", scipy.__version__)
```

```python
a = np.array([10, 20, 30, 40, 50])

print("Array     =", a)
print("Mean      =", np.mean(a))
print("Sum       =", np.sum(a))
print("Std. dev. =", round(np.std(a), 4))
```
