```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["figure.figsize"] = (7, 4)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3

df = pd.read_csv("popular_anime.csv")
df["year"] = pd.to_datetime(df["aired_from"], errors="coerce", utc=True).dt.year
print("Shape :", df.shape)
```

```python
# ---------- Line plot : titles released per year ----------
per_year = df[df["year"].between(1980, 2024)]["year"].value_counts().sort_index()

plt.figure()
plt.plot(per_year.index, per_year.values, color="#2a6fdb", linewidth=2)
plt.title("Number of Anime Titles Released Per Year")
plt.xlabel("Year")
plt.ylabel("Number of titles")
plt.tight_layout()
plt.show()
```

```python
# ---------- Bar chart : titles by type ----------
by_type = df["type"].value_counts()

plt.figure()
plt.bar(by_type.index, by_type.values, color="#4c9f70", edgecolor="black")
plt.title("Number of Titles by Content Type")
plt.xlabel("Type")
plt.ylabel("Number of titles")
for i, v in enumerate(by_type.values):
    plt.text(i, v + 200, str(v), ha="center", fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# ---------- Horizontal bar chart : highest rated titles ----------
top10 = df.dropna(subset=["score"]).nlargest(10, "score")[["name", "score"]]
short = [n[:28] + "..." if len(n) > 28 else n for n in top10["name"]]

plt.figure(figsize=(7, 4.5))
plt.barh(short[::-1], top10["score"][::-1], color="#e07a5f", edgecolor="black")
plt.title("Ten Highest Rated Anime Titles")
plt.xlabel("Score")
plt.xlim(8.5, 9.5)
plt.tight_layout()
plt.show()
```

```python
# ---------- Histogram : distribution of the score ----------
plt.figure()
plt.hist(df["score"].dropna(), bins=40, color="#7b68ee", edgecolor="white")
plt.axvline(df["score"].mean(), color="red", linestyle="--", linewidth=2,
            label="Mean = %.2f" % df["score"].mean())
plt.title("Distribution of Anime Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ---------- Pie chart : share of content rating ----------
ratings = df["rating"].value_counts().head(5)
labels = [r.split(" - ")[0] for r in ratings.index]

plt.figure(figsize=(6, 5))
plt.pie(ratings.values, labels=labels, autopct="%1.1f%%", startangle=100,
        colors=["#4c9f70", "#e07a5f", "#2a6fdb", "#f2cc8f", "#9c89b8"],
        wedgeprops={"edgecolor": "white"})
plt.title("Share of Each Content Rating Category")
plt.tight_layout()
plt.show()
```

```python
# ---------- Scatter plot : votes against score ----------
s = df.dropna(subset=["score", "scored_by"]).sample(4000, random_state=7)

plt.figure()
plt.scatter(s["scored_by"], s["score"], s=10, alpha=0.3, color="#2a6fdb")
plt.xscale("log")
plt.title("Score against Number of Votes")
plt.xlabel("Number of votes  (log scale)")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

```python
# ---------- Box plot : score distribution by type ----------
types = ["TV", "Movie", "OVA", "ONA", "Special"]
groups = [df[df["type"] == t]["score"].dropna() for t in types]

plt.figure()
plt.boxplot(groups, labels=types, patch_artist=True,
            boxprops={"facecolor": "#a8dadc"},
            medianprops={"color": "red", "linewidth": 2})
plt.title("Score Distribution Across Content Types")
plt.xlabel("Type")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

```python
# ---------- Multiple plots using subplots ----------
fig, ax = plt.subplots(2, 2, figsize=(11, 7))

ax[0, 0].plot(per_year.index, per_year.values, color="#2a6fdb")
ax[0, 0].set_title("Titles Per Year")

ax[0, 1].bar(by_type.index, by_type.values, color="#4c9f70")
ax[0, 1].set_title("Titles by Type")

ax[1, 0].hist(df["score"].dropna(), bins=30, color="#7b68ee", edgecolor="white")
ax[1, 0].set_title("Score Distribution")

ax[1, 1].boxplot(groups, labels=types, patch_artist=True,
                 boxprops={"facecolor": "#a8dadc"})
ax[1, 1].set_title("Score by Type")

for a in ax.flat:
    a.grid(alpha=0.3)
fig.suptitle("Anime Dataset - Summary Dashboard", fontsize=13)
plt.tight_layout()
plt.savefig("anime_dashboard.png", dpi=100)
plt.show()

print("Figure saved as anime_dashboard.png")
```
